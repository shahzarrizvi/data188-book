## Background
So far, we have been working with parametric tests for a single population. Our hypothesis tests have focused on some parameter (e.g. mean) in one underlying population. But in practice, rarely do we care about whether the mean of the population is some exact value or not. It is much more common to be working with two populations, where we get a sample from each and compare the sample behavior to conclude if there is some underlying population difference.


## Testing population means
Assume we have two **independent** samples $X_1, X_2, ..., X_n$ and $Y_1, Y_2,..., Y_m$, both iid. 

Our null hypothesis is that these two samples are drawn from two underlying populations with the same mean (not necessarily the same distribution). 
$$H_0: \mu_{X} = \mu_{Y}$$

A natural test statistic to consider is $\bar{X} - \bar{Y}$, the difference between the sample means. Depending on our alternative hypothesis, we may need to make slight modifications (e.g. taking absolute value), but at the heart of it is the difference in sample means. 

Regardless of the underlying distributions of $X$ and $Y$, the mean and variance of our statistic are:
$$E(\bar{X} - \bar{Y}) = \mu_{X} - \mu_{Y}$$
$$Var(\bar{X} - \bar{Y}) = \frac{\sigma_{X}^{2}}{n} + \frac{\sigma_{Y}^{2}}{m}$$

We want to measure how far our statistic is from 0, so we construct a new statistic $T$ that measures this in standardized units.
$$ T = \frac{(\bar{X} - \bar{Y}) - 0}{\sqrt{\frac{\sigma_{X}^{2}}{n} + \frac{\sigma_{Y}^{2}}{m}}} $$

Notice that $T$ is equivalent to the standardized $\bar{X} - \bar{Y}$ under the null hypothesis. The mean of $\bar{X} - \bar{Y}$ is 0 because $E(\bar{X})$ = $E(\bar{Y})$ under $H_{0}$.




### **Distribution of $T$**
To compute $T$ and figure out its distribution, we need to either **know** or **estimate** $\sigma_{X}$ and $\sigma_{Y}$. We have the following results:

**Case 1.** If the X's and Y's are normal, and $\sigma_{X}$ and $\sigma_{Y}$ are *known*, then $T $ is standard normal.

**Case 2.** If the X's and Y's are normal, but $\sigma_{X}$ and $\sigma_{Y}$ are *unknown*
<br> We need to create another statistic with estimates for $\sigma_{X}$ and $\sigma_{Y}$.
$$ T_2 = \frac{(\bar{X} - \bar{Y}) - 0}{\sqrt{\frac{S_{X}^{2}}{n} + \frac{S_{Y}^{2}}{m}}} \sim t_{\min(n-1, m-1)} $$
<br> where $S_{X}^{2}$ and $S_{Y}^{2}$ are variance estimates from our two samples. The proof of this distribution is complicated (see Welch's test), but **conservatively** it is a t-distribution with $min(n-1, m-1)$ degrees of freedom. 

**Case 3.** If the X's and Y's are normal, $\sigma_{X}$ and $\sigma_{Y}$ are *unknown*, but we know that $\sigma_{X} = \sigma_{Y}$. 
<br> We will use a pooled sample variance as a common estimate of population variance $\sigma^{2}$:
$$ S_{pooled}^{2} = \frac{(n-1)S_{X}^{2} + (m-1)S_{Y}^{2}}{n+m-2} $$
<br> Our new statistic becomes
$$ T_{3} = \frac{(\bar{X} - \bar{Y}) - 0}{\sqrt{S_{\rm pooled}^{2} \left(\frac{1}{n} + \frac{1}{m}\right)}} \sim t_{n+m-2}$$
<br> Plugging in the definition of $S_{\rm pooled}^{2}$ and simplifying shows that $T_3$ is **exactly** a t-distribution with $n+m-2$ degrees of freedom. 

Derivation of $S_{\rm pooled}^{2}$: 
<br> We use chi-squared distribution properties to suggest weights on our variance estimates, $S_{X}^{2}$ and $S_{Y}^{2}$. In Data 140, we derived the following property about our variance estimates: 
$$ S_{X}^{2} = \frac{1}{n-1} \sum_{i=1}^{n} (X_i - \bar{X})^{2} $$
$$ (n-1) S_{X}^{2} = \sum_{i=1}^{n} (X_i - \bar{X})^{2}$$
$$(n-1) \frac{S_{X}^{2}}{\sigma^{2}} = \sum_{i=1}^{n} \left(\frac{X_i - \bar{X}}{\sigma}\right)^{2} \sim \chi^{2}_{n-1} $$
<br> Similarly for $Y$,
$$ (m-1) \frac{S_{Y}^{2}}{\sigma^{2}} \sim \chi^{2}_{m-1}$$

<br> By additive property of independent chi-squared distribution, we have
$$ (n-1) \frac{S_{X}^{2}}{\sigma^{2}} + (m-1) \frac{S_{Y}^{2}}{\sigma^{2}} \sim \chi^{2}_{n+m-2}$$
<br> So to combine the two variance estimates into a common estimate, we essentially weigh them by their chi-squared degrees of freedom. 

**Note that the three cases above make no assumption on the sample sizes $n$ and $m$.**

**Case 4.** If $n$ and $m$ are large, CLT says
$$ \frac{(\bar{X} - \bar{Y}) - 0}{\sqrt{\frac{\mathord{?}_{X}^{2}}{n} + \frac{\mathord{?}_{Y}^{2}}{m}}} \rightarrow \mathcal{N}(0, 1)$$ 
<br> where $\mathord{?}$ can be either $\sigma^{2}, \hat{\sigma}^{2}$ or $S^{2}$ because they are asymptotically equivalent.

## The Permutation Test

Let's say we have two samples. $X_1, X_2,..., X_n$, which are iid from a (not necessarily specified) distribution $F_X$ and $Y_1, Y_2, ..., Y_m$ which are iid from a (again not necessarily specified) distribution $F_Y$. We also need to make a couple assumptions for everything to work out nicely, like our distributions have finite means. 

Our null hypothesis is that these two samples are drawn from the same distribution, or $$H_0: F_X = F_Y$$

As is typical, we won't consider our alternate hypothesis when constructing our test. Let's let our test statistic be $T:=\overline{X}-\overline{Y}.$ If we have large samples or we assume gaussian $F_X$, then T would be normal. However, without these assumptions there is little we can say about the distribution of $T.$

So we turn to our null hypothesis to simulate a distribution of $T$. If $F_X = F_Y$, then we essentially just have a single sample of size $n+m$ from this distribution. Let's call this sample $$\mathbb{X} = \{X_1, X_2, ..., X_n, Y_1, Y_2,...,Y_m\}.$$

We can then permute $\mathbb{X}$, which we will denote as $\pi(\mathbb{X})$, and take the first $n$ observations in this sample to be our $F_X$ sample and the rest to be our $F_Y$ sample. Our test statistic $$T(\pi(\mathbb{X})) = \text{mean(The first }n\text{)} - \text{mean(The last }m\text{)}.$$ If we compute all $(n+m)!$ possible permutations, we will have our true distribution of T, conditioned on $\mathbb{X}$. In practice, simulating a large number of permutations is enough to approximate this distribution.

We reject the null hypothesis if the test statistic of our original data lie in the tails of our simulated distribution. The level of this test is therefore $P(R \mid \mathbb{X}),$ where R is our rejection region. However, our normal level $\alpha = P(R),$  does not depend on our sample. However, these are equivalent. 
$$
\begin{align*} 
    E_{H_0}(I(\text{Reject }H_0)) &= E_{H_0}\Big(E_{H_0}\big(I(\text{Reject }H_0 \mid \mathbb{X})\big)\Big)\\
    &= E_{H_0}\big(P_{H_0}(\text{Reject}) \mid \mathbb{X}\big)\\
    &= P_{H_0}(R)\\
    &=\alpha
\end{align*}
$$

It is also important to point out the limitations of this test. Means are inherently sensitive to outlier data and therefore this test may not work well when the data have outliers. Similarly, accepting the null for this specific test statistic does not necessarily mean the two distributions are the same, but rather that they have similar means. 


As an aside, a domain where the permutation test is almost certainly invalid is in time series data, because our sample isn't equally likely if we permute it. However, some researchers from Stanford extended the [permutation test to the time series domain](https://arxiv.org/pdf/2009.03170.pdf). 


## The Wilcoxon Rank Sum Test

Let's see if we can handle the permutation test's sensitivity to outliers. Here, we have a similar setup to the permutation test. 
$X_1, X_2,..., X_n$, are iid from a (not necessarily specified) distribution $F_X$ and $Y_1, Y_2, ..., Y_m$ are iid from a (again not necessarily specified) distribution $F_Y$. Again, $H_0: F_X = F_Y.$

To help address the permutation test's sensitivity to outliers we can use the rankings of our data instead of their actual values. So if our data look like this:
$$x_1, x_2, x_3, y_1 = 2,4,1,-100$$
Their ranks would be (with $1$ being the minimum and $n+m$ being the maximum):
$$x_1, x_2, x_3, y_1 = 3,4,2,1.$$
For simplicity, let's assume that our $F_X$ is continuous so we don't have to deal with ties. This test does work without this assumption, but the algebra is a bit messier. 

If our null hypothesis is true, our sample $X$s should look like a simple random sample of our aggregate rankings dataset. 

Our test statistic is $W := \text{sum(ranks of }X\text{s)}$. If W is low or high, then we reject the null. To formalize this, we need to draw a bit on our knowledge of simple random samples from Data 140. Under our null hypothesis, each of our $n$ variables has expected value $\frac{n+m+1}{2},$ or the average rank (which is Uniform from 1 to N = n+m). Therefore, by applying linearity of expectation, we get $$E(W) = n\cdot\frac{n+m+1}{2}$$

For the variance of our test statistic, let's first consider the case of an infinite-sized combined sample. 
$$Var(W_\inf) := n\cdot\frac{(n+m)^2-1}{12} = n\frac{(n+m-1)(n+m+1)}{12}$$
Here, we use the fact independent RVs' variances are additive. However, we have ignored the fact that our sample occurs from a finite population, so we must use the finite population correction factor: $\frac{(n+m)-m}{(n+m)-1}.$ This gives us the variance of our test statistic.
$$Var(W) = n\cdot\frac{(m+n+1)(m+n-1)}{12}\cdot\frac{n}{n+m-1} = \frac{mn(n+m+1)}{12}.$$

Chebyshev's bound will give you a rough range where the rank sum statistic is overwhelmingly likely to fall.

Or you can simulate the rank sum statistic under the null hypothesis and compare your observed rank sum to the simulated distribution.

It turns out that if $n$ and $m$ are large then the distribution of $W$ is approximately normal. In that case all inference can be done based on this curve, using the expectation and variance calculated above.
