# Goodness of Fit

That's statistics language for how well the data and the model fit each other. We are going to move from testing whether two underlying distributions are the same to whether a single i.i.d. sample is drawn from a specified distribution. 

First, a recap of a calculation done earlier:

Suppose we have data  $X_1,X_2,...,X_n$  i.i.d normal $(\mu,\sigma^2)$ with unknown $\mu$ and known $\sigma$. We will perform a generalized likelihood ratio test of two hypotheses:$$H_0: \mu=\mu_0$$ $$H_1: \mu\neq\mu_0$$.

Then the generalized likelihood ratio is
$$\begin{align*}
\Lambda &=\frac{\displaystyle\prod_{i=1}^{n}\frac{1}{\sqrt{2\pi}\sigma}\exp\left\{-\frac{1}{2\sigma^2}(X_i-\mu_0)^2\right\}}{\displaystyle\prod_{i=1}^{n}\frac{1}{\sqrt{2\pi}\sigma}\exp\left\{-\frac{1}{2\sigma^2}(X_i-\bar{X})^2\right\}}\\ 
&=\exp\left\{-\frac{1}{2} \cdot n\left(\frac{\bar{X}_n - \mu_0}{\sigma} \right)^2 \right\} \end{align*}$$
by algebra. Thus

$$
-2\log(\Lambda) = \left( \frac{\bar{X}_n - \mu_0}{\sigma/\sqrt{n}} \right)^2
$$

The right hand side is the square of a standard normal. As we saw in Data 140, this has the gamma $(1/2, 1/2)$ distribution, also known as the the $\chi^2$ distribution with $1$ degree of freedom.

It is sometimes true (think CLT) that a result that is true for normal data for all sample sizes can turn out to be approximately true for all data if the sample size is large. The fact below, which we will not prove, is another instance of this.

### Generalized Likelihood Ratio Test

Suppose we have data $X_1,X_2,...,X_n$ i.i.d $f(x \mid \theta)$ where $\theta$ belongs to some set $S$. Note that $\theta$ might be a vector of parameters.

Suppose we are testing $H_0: \theta = \theta_0$ versus $H_1: \theta \neq \theta_0$ for some fixed $\theta_0 \in S$. Assume nice conditions on $f$ and $S$. Then under $H_0$, $-2\log{\Lambda}\xrightarrow[n\rightarrow\infty]{d}\chi^2_k$, for $k=$ "number of free parameters".

### Example: The Multinomial Model

We will test whether or not our data are i.i.d. draws from a categorical distribution with a finite number of categories.

$H_0$: Sample is i.i.d from a known categorical distribution with $c$ categories

$H_1$: Sample is i.i.d from some different distribution on the $c$ categories

Let the proportions in the null distribution be $\theta_{0, i}$ for $1 \le i \le c$, and note that $\sum_{i=1}^c \theta_{0,i} = 1$. Let the sample size be $n$. 

For $1 \le i \le c$, let $E_i$ be the expected frequency in Category $i$ under $H_0$. Then $E_i = n\theta_{0, i}$ under $H_0$.

For $i = 1,2...,c$, let $O_i$ be the observed frequency (that is, count) in Category $i$. Recall that the MLE of $\theta_i$, the true probability that a draw is in Category $i$, is the observed proportion in that category. That is, $\hat{\theta}_i = O_i/n$.

Thus the generalized likelihood ratio is

$$\begin{aligned}\Lambda &= \frac{\frac{n!}{O_1!O_2!\cdots O_c!}\prod_{i=1}^{c}(\theta_{0,i})^{O_i}}{\frac{n!}{O_1!O_2!\cdots O_c!}\prod_{i=1}^{c}(\hat{\theta})^{O_i}}\\
&= \frac{\prod_{i=1}^{c}(\theta_{0,i})^{O_i} }
{ {\prod_{i=1}^c(O_i/n})^{O_i}}\\
&= \frac{\prod_{i=1}^{c}(n\theta_{0,i})^{O_i} }
{ {\prod_{i=1}^c(O_i})^{O_i}}\\
&= \frac{\prod_{i=1}^{c} {E_i}^{O_i}}{\prod_{i=1}^{c}{O_i}^{O_i}}\\
&=\prod_{i=1}^{c}\left(\frac{E_i}{O_i}\right)^{O_i}
 \end{aligned}$$

So
$$-2\log\Lambda = 2\sum_{i=1}^{c}{O_i}\log\frac{O_i}{E_i}$$

By our theorem on the generalized likelihood ratio test, under $H_0$ the distribution of this statistic converges to $\chi^2_{c-1}$ as the sample size gets large.

The degrees of freedom are $c-1$ because the proportions in the $c$ categories have to add up to $1$. So there are only $c-1$ free parameters. This isn't intended as a proof but simply as a way to understand and remember the degrees of freedom.

### Pearson's Approximation of the Statistic

Pearson showed that statistic above can be approximated by the following expression for large a sample size. This approximation made it easy to conduct $\chi^2$ test without access to computers or calculators. These days, it's just as easy to stick with the original expression.

$$
-2\log \Lambda ~=~ 2\sum_{i=1}^{c}{O_i}\log\frac{O_i}{E_i} ~\approx~ \sum_{i=1}^c \frac{ (O_i - E_i)^2 }{E_i} 
$$

### Chi-Squared Statistic or Total Variation Distance?

In Data 8 we used total variation distance to measure the distance between two categorical distributions. We were therefore able to use it as a test statistic to decide between $H_0$ and $H_1$. This method works regardless of the sample size. However, since the null distribution of the TVD isn't easy to find, we need to simulate data under $H_0$ every time. 

The chi-squared test is computationally simpler (no simulation required) but requires large samples for the chi-squared approximation to work.

### Example: Do the Data Follow a Poisson Distribution?

Let $X_1, X_2, \ldots, X_n$ be i.i.d. and suppose we are trying to decide whether the distribution of $X_i$ is Poisson.

**Case 1: Known Parameter $\lambda$**

The hypotheses are 
- $H_0$: The data are i.i.d. Poisson $(\lambda)$.
- $H_1$: The data are i.i.d. Poisson but with a parameter different from the $\lambda$ specified in $H_0$.

**Step 1:** Truncate the sampled data to have $c$ categories $0, 1, \dots c-1$. This truncation reduces the Poisson to a distribution on a finite number of values. If you choose $c$ to be big enough, this will not cause much of a problem.

**Step 2:** For all $i\in\{0, 1,2, \dots, c-1\}$, calculate $E_i$, the expected count under $H_0$ in Category $i$ under the Poisson ($\lambda$) distribution. For $1 \le i \le c-2$, we have $E_i = \displaystyle n \cdot e^{-\lambda}\frac{\lambda^i}{i!}$.
Because of the truncation, 

$$ E_{c-1} =  n\left(1- \sum_{i=1}^{c-2} e^{-\lambda}\frac{\lambda^i}{i!}\right)
$$

**Step 3:** Calculate the chi-squared statistic (either one) developed above. Since there are $c$ categories, compare the statistic with the $\chi^2$ distribution with $c-1$ degrees of freedom. 

**Case 2: Unknown parameter $\lambda$**

Start with:

**Step 0:** Estimate $\lambda$ from the data, using the MLE. In the case of a Poisson sample, the MLE is $\hat{\lambda} = \bar{X}_n$, the sample mean. This was a Data 140 exercise.

Then run Steps 1 and 2 above, using $\hat{\lambda}$ in place of $\lambda$. Replace Step 3 with:

**New Step 3:** Calculate the chi-squared test statistic and compare it with the $\chi^2_{c-1-1}$ distribution. The extra $-1$ in the degrees of freedom comes from the fact that we estimated $\lambda$ from the data, which means there is less deviation between observed counts and expected counts than there would have been if we were just drawing from a fixed distribution. See the worksheet for a simulation study.

## Continuous Case: Kolmogorov-Smirnov Test

The test above is used to decide whether our data come from a specified discrete distribution. We now move to the continuous case: testing whether the data come from a specified continuous cdf. 

Let $X_1, X_2,..., X_n$ be iid and let their common cdf be denoted $F_X$. Our null hypothesis is $H_0: F_X = F$, where $F$ is a fully specified cdf (that is, it is a known function with no unknown parameters to estimate).

We will work under a critical assumption: $F$ is continuous and increasing so that $F^{-1}$ exists.

A good approach is to see how the empirical cdf of the data differs from $F$. Recall that he empirical cdf of the data is defined as follows:

$$F_n(x) = \frac{1}{n}\sum_{i=1}^{n} I(X_i \le x) ~~~~~~~~~ \text{ for all } x$$

Thus for each fixed $x$, $F_n(x)$ is the proportion of data points that are at or below $x$. 

The **Kolmogorov-Smirnov test statistic** is the *biggest difference between $F_n$ and $F$ over all points $x$*:

$$D_n = \sup_{x\in \mathbb{R}}|F_n(x) - F(x)|$$

Here $\sup$ stands for *supremum*. That's like a maximum taken over the entire real line. In general, such a "maximum" need not be attained. However, since $F$ is increasing,
the largest difference between $F$ and $F_n$ must occur at one of the jumps of $F_n$. Therefore the supremum is actually a maximum of $n$ values.

Under the null, $F_n$ should approximate $F$ if $n$ is large. So $D_n$ should become smaller if $n$ gets larger. This should be believable though its formal proof is technical and beyond our scope.

The good news is that $D_n$ is easy to calculate. The bad news is that its null distribution appears to depends a lot on exactly which $F$ is specified in $H_0$. 

But appearances are deceptive. We will now see how the null distribution of $D_n$ does not depend on $F$. 

Before we do that, take a moment to think about what this result means. It tells us that whether we are trying to see if our data are standard normal or exponential $(1/2)$, the null distribution of the Kolmogorov-Smirnov test statistic is the same. It depends only on the sample size. We don't need to work out one distribution when the null specifies the standard normal, another distribution when the null specifies exponential $(1/2)$, and so on.

The key observation is that we are looking at the biggest difference between pairs of probabilities. All the action is on the unit interval. We will use that to our advantage.

In everything that follows, keep in mind that we are working under $H_0$.

Then our sample $X_1, X_2, \ldots, X_n$ is i.i.d. $F$. It is defined on some probability space about which we don't know anything. To study the null distribution of $D_n$, we will construct another sequence $X_1^*, X_2^*, \ldots, X_n^*$ on a different probability space so that $X_1^*, X_2^*, \ldots, X_n^*$ is i.i.d. $F$ (just like our sample) and the new space is such that we can see how $D_n^*$ behaves.

If $(X_1, X_2, \ldots, X_n) \stackrel{d}{=} (X_1^*, X_2^*, \ldots, X_n^*)$, then $D_n \stackrel{d}{=} D_n^*$. So let's construct the new sample and study $D_n^*$.

Take a space on which you have $U_1, U_2, ...$ that are iid uniform $(0,1)$ random variables. For $i \ge 1$ let $X_i^* = F^{-1}(U_i)$. Recall from Data 140 that $X_i^*$ has cdf $F$. The $X_i^*$s are independent of each other since the $U_i$'s are independent of each other. Thus $X_1^*, X_2^*, \ldots, X_n^*$ are i.i.d. $F$.

Now define
$$
D_n^* = \sup_{x \in \mathbb{R}} \vert F_n^*(x) - F(x)| ~~~ \text{ where } ~~~
F_n^*(x) = \frac{1}{n} \sum_{i=1}^n I(X_i^* \le x)
$$

Since $F$ is increasing, $X_i^* \le x \iff F(X_i^*) \le F(x)$. By the definition of $X_i^*$ we have $F(X_i^*) = U_i$. Therefore

$$
F_n^*(x) ~=~ \frac{1}{n} \sum_{i=1}^n I(X_i \le x)
~=~ \frac{1}{n} \sum_{i=1}^n I(U_i \le F(x)) ~=~ G_n(F(x))
$$
where $G_n$ is the empirical cdf of $U_1, U_2, \ldots, U_n$. Thus

$$
\begin{align*}
D_n^* &= \sup_{x \in \mathbb{R}} \vert F_n^*(x) - F(x)| \\
&= \sup_{x \in \mathbb{R}} \vert G_n(F(x)) - F(x) \vert \\
&= \sup_{0 \le u \le 1} |G_n(u)-u|
\end{align*}
$$

The last step above follows because as $x$ ranges over all of $\mathbb{R}$, $F(x)$ ranges over the unit interval.

Thus $D_n^* = \sup_{0 \le u \le 1} |G_n(u)-u|$ which doesn't involve $F$.

Back to our original sample $X_1, X_2, \ldots, X_n$: Under $H_0$, we have $D_n \stackrel{d}{=} D_n^*$. Hence, under $H_0$, we have $D_n \stackrel{d}{=} \sup_{0 \le u \le 1} |G_n(u)-u|$ where $G_n$ is the empirical cdf of $n$ i.i.d. uniform $(0, 1)$ random variables.

You have a few options for finding this distribution:

- By simulation: Do the following repeatedly and plot the histogram of the results:
    - Draw $n$ times independently from the uniform $(0, 1)$ distribution and calculate the resulting value of $\sup_{0 \le u \le 1} |G_n(u)-u|$.
- By Python: See the worksheet.
- By probability theory: Take Stat 205. Suitably normalized, $\sup_{0 \le u \le 1} |G_n(u)-u|$ has a lovely connection to the Brownian Bridge.
