## Testing Hypotheses

Here is a quick recap of terminology, starting with two views of the world from which the data were generated (these are also called models):

- Null hypothesis $H_0$
- Alternative hypothesis $H_1$

Let the data be $\mathbb{X}$. The goal of hypothesis testing is to determine which model is better supported by the data.

If the test finds that the data are more compatible with $H_1$ than with $H_0$, then the test will reject the null hypothesis and conclude $H_1$. Otherwise it will conclude that the data are compatible with $H_0$.

## Decision Rules

Formally, a test or *decision rule* is defined as a set of conditions on the data that determine whether we will accept or reject our null hypothesis. This decision is based on the sample; typically, it is based on a test statistic. 

**Decision Rule**: Reject $H_0$ if the data fall in a region $R$, called the *rejection region*.

For example, the rejection region might look like "sample mean is greater than 25.5".

The data can fall either into the rejection region, $R$, or acceptance region, $A = R^c$. The rejection region is the set of values of the data where the decision rule chooses to reject the null hypothesis in favor of the alternative. The *acceptance region* is the complementary set where we choose to stay with the null hypothesis. 

Any test can therefore be thought of as a rejection region, as follows:

- If $\mathbb{X} \in R$, the test concludes $H_1$.
- If $\mathbb{X} \in A$, the test concludes $H_0$.

## Level and Power
The conclusion of such a test can be wrong. Here's a table we've all seen before. This is the familiar confusion matrix from Data 100, a convenient way to represent errors in our decision-making process.

|                    | **Test: $H_0$**      |**Test: $H_1$**     |
| -------------------|----------------------|--------------------| 
| **Truth: $H_0$**   | True Negative        | False Positive     |
| **Truth: $H_1$**   | False Negative       | True Positive      |

Let's populate the matrix with some probabilities. **The discussion below assumes that probabilities can be calculated under both hypotheses.** Keep in mind that if the alternative is something like, "The jury panel was not selected at random," then in the frequentist world there are no probabilities unde that hypothesis.

|                    | **Test: $H_0$**       |  **Test: $H_1$**    |
| -------------------|-----------------------|---------------------| 
| **Truth: $H_0$**   | $$P_{H_0}(A)$$        | $$P_{H_0}(R)$$      |
| **Truth: $H_1$**   | $$P_{H_1}(A)$$        | $$P_{H_1}(R)$$      |

The probabilities are of the events $A$ or $R$, over the ground truth, $H_0$ or $H_1$. Note that the each row sums to 1. But the columns don't.

In standard notation, the table above becomes:

|                    | **Test: $H_0$**         |  **Test: $H_1$** |
| -------------------|-------------------------|------------------| 
| **Truth: $H_0$**   | $$1 - \alpha$$          | $$\alpha$$       |
| **Truth: $H_1$**   | $$\beta$$               | $$1 - \beta$$    |

- $\alpha$ is known as the *level* of the test. It's the probability "Type I error": concluding $H_1$ when $H_0$ is fact true. In the familiar Data 8 hypothesis testing world, it's also known as the $p$-value threshold. Since the level is an error probability, it is typically chosen to be small.

- $\beta$ is known as the probability of "Type II error": concluding $H_0$ when $H_1$ is true.

- $1-\beta$ is known as the *power* of the test. It's the probability of a correct conclusion: concluding $H_1$ when $H_1$ is true. Since the power is the chance of a correct answer, we would like it to be high.

### Key Observation
Both level and power are probabilities of concluding $H_1$, that is, they are probabilities of the rejection region. 
- The level is the probability of the rejection region under $H_0$. 
- The power is the probability of the same rejection region under $H_1$.

## Simple and Composite Hypotheses

A *simple hypothesis* is a completely specified probability model. There are no unknowns in such a model. You can simulate data under it. For example, $p = 0.5$ is a simple hypothesis, as is $p = 0.2$ or $\mu = 10$.

A *composite hypothesis* is one that doesn't completely specify the model. For example, $p \neq 0.5$ is composite.

**Example:** Suppose we have two simple hypotheses about the chance that a coin lands heads:

- $H_0: p = 0.5$
- $H_1: p = 0.7$.

Now suppose our data $\mathbb{X}$ are the results of 100 tosses of the coin, and suppose our decision rule is:

- If the number of heads is at least $60$ then reject $H_0$. Otherwise stay with $H_0$.

This test is defined by its rejection region $R$: "number of heads $\ge 60$".

The level $\alpha$ of this test is the probability that a binomial $(100, 0.5)$ random variable is at least $60$. Since the binomial $(100, 0.5)$ distribution is approximately normal with mean $50$ and SD $5$, we get $\alpha \approx 2.5\%$.

The power of the test is the chance that a binomial $(100, 0.7)$ random variable is at least $60$. The distribution has mean $70$ and SD less than $5$ (because $p(1-p)$ is maximized at $p=0.5$). So the power will be large.

## Likelihood Ratio Test

Let $f_0$ be the likelihood of the data under $H_0$, and let $f_1$ be the likelihood under $H_1$. This is the same concept of likelihood that we examined when we studied the MLE.

### Likelihood Ratio
Let $H_0$ and $H_1$ both be simple hypotheses let $R$ be the rejection region of our test. Define ${\rm LR}(\mathbb{X})$ to be the likelihood ratio of the data $\mathbb{X}$:

$$
{\rm LR}(\mathbb{X}) = \frac{f_0(\mathbb{X})}{f_1(\mathbb{X})}
$$

### Likelihood Ratio Test (LRT)

A Likelihood Ratio Test (LRT) rejects the null hypothesis if ${\rm LR}(\mathbb{X}) < c$ for some threshold $c > 0$.

This is motivated by the fact that the likelihood ratio is small when the data are more likely under the alternative hypothesis than under the null. 

The likelihood ratio is often a complicated statistic. But all we want is to understand when it's small. So when we construct likelihood ratio tests, we will try to find the simplest possible way to express the condition "the likelihood ratio is small".

## Neyman-Pearson Lemma

**Lemma:** Let $\alpha$ be the level of the LRT. Any other test with level $\le \alpha$ will have a power no greater than the power of the LRT.

The lemma implies that the likelihood ratio test is the most powerful test of two simple hypotheses, among all tests of its level or lower.

To establish the Neyman-Pearson Lemma, we will start with some observations.

First, consider any two tests $\tau$ and $T$. Let $R_\tau$ and $A_\tau$ respectively be the rejection and acceptance regions of the test $\tau$. And let $R_T$ and $A_T$ be the rejection and acceptance regions of the test $T$.

Then the outcome space splits four ways.

|                    |                      |
| -------------------|----------------------| 
| $$R_\tau R_T$$     | $$R_\tau A_T$$       |
| $$A_\tau R_T$$     | $$A_\tau A_T$$       |

In this partition:

- $R_\tau$ is the union of the events in the first row.
- $R_T$ is the union of the events in the first column.

This gives us **Observation 1:** For any probability distribution $P$,

$$
P(R_\tau) - P(R_T) ~ = ~ P(R_\tau A_T) - P(A_\tau R_T)
$$

Now let $L$ be the likelihood ratio test, and let $R_L$ and $A_L$ be its rejection and acceptance regions respectively.

Note that if $\mathbb{X} \in R_L$ then $f_0(\mathbb{X}) < cf_1(\mathbb{X})$. And if $\mathbb{X} \in A_L$ then $f_0(\mathbb{X}) \ge cf_1(\mathbb{X})$.

We thus have **Observation 2:** 
- For any set $S \subseteq R_L$, we have $P_0(S) < cP_1(S)$.
- For any set $S \subseteq A_L$, we have $P_0(S) \ge cP_1(S)$.

### Proof of the Neyman-Pearson Lemma

Now let $T$ be a competing test of level no greater than the level of $L$. Our goal is to show that the power of $T$ is no greater than the power of $L$. 

First **note** the following by Observation 2:

- Since $R_L A_T \subseteq R_L$, we have $P_1(R_L A_T) > \displaystyle \frac{1}{c} P_0(R_L A_T)$.
- Since $A_L R_T \subseteq A_L$, we have $P_1(A_L R_T) \leq \displaystyle \frac{1}{c} P_0(A_L R_T)$.

Now compare the powers of the two tests.

\begin{align}
P_1(R_L) - P_1(R_T) &= P_1(R_L A_T) - P_1(A_L R_T) & \text{by Observation 1}\\
&\geq \frac{1}{c} P_0(R_L A_T) - \frac{1}{c} P_0(A_L R_T) & \text{by the note above}\\
&= \frac{1}{c}\left( P_0(R_L A_T) - P_0(A_L R_T) \right)\\
&= \frac{1}{c}\left( P_0(R_L) - P_0(R_T) \right) & \text{by Observation 1} \\
&\ge 0
\end{align}

because the level of $R_L$ is at least as large as the level of $R_T$ by assumption.

Thus $P_1(R_L) \ge P_1(R_T)$. That is, the power of the likelihood ratio test is at least as large as the power of the competing test.

## Example: LRT for the Normal Mean ###

Suppose we have data $X_1, X_2, ..., X_n$ i.i.d Normal $(\mu, 1)$, and we want to perform a likelihood ratio test of two simple hypotheses:

- $H_0: \mu = \mu_0$
- $H_1: \mu = \mu_1$ 


Then
$$
\begin{align*}
{\rm LR}(\mathbb{X}) &= \frac{f_0(\mathbb{X})}{f_1(\mathbb{X})}\\
&= \frac{\left(\frac{1}{\sqrt{2\pi}}\right)^n \cdot \exp\left\{-\frac{1}{2} \sum^n_{i=1} (X_i - \mu_0)^2\right\}}
{\left(\frac{1}{\sqrt{2\pi}}\right)^n \cdot \exp\left\{-\frac{1}{2} \sum^n_{i=1} (X_i - \mu_1)^2\right\}} \\
&= \exp\left\{-\frac{1}{2}\left(\sum^n_{i=1} (X_i - \mu_0)^2 - \sum^n_{i=1} (X_i - \mu_1)^2\right)\right\} \\
&= \exp\left\{-\frac{1}{2} \sum^n_{i=1} \Big(2X_i - (\mu_0+\mu_1)\Big)\Big(\mu_1-\mu_0\Big)\right\} \\
&= \exp\left\{-(\mu_1 - \mu_0)\cdot n\bar{X}_n + \frac{1}{2}(\mu_1^2-\mu_0^2)n\right\}
\end{align*}
$$

Therefore ${\rm LR}(\mathbb{X})$ is a function of the sample mean $\bar{X}_n$. Since the rejection region is based on ${\rm LR}(\mathbb{X})$ being small, let's to express that condition as simply as possible based on $\bar{X}_n$.

First note that the factor $\displaystyle \exp\left\{-\frac{1}{2}(\mu_1^2-\mu_0^2)n\right\}$ is a constant, and hence

$$
{\rm LR}(\mathbb{X}) \text{ is small } ~ \iff ~ \exp\left\{-(\mu_1 - \mu_0)\cdot n\bar{X}_n\right\} \text { is small}
$$

Thus:

- If $\mu_1 > \mu_0$, then ${\rm LR}(\mathbb{X})$ is small when $\bar{X}_n$ is large.
- If $\mu_0 > \mu_1$, then ${\rm LR}(\mathbb{X})$ is small when $\bar{X}_n$ is small.


This makes sense. If our alternative hypothesis proposes a larger $\mu$ than that in our null, then a large sample mean would sway us toward the alternative. On the other hand, if our alternative hypothesis proposes a smaller $\mu$ than that in our null, then a small sample mean would sway us toward the alternative. 
    
Finally, to define precisely what "large" or "small" mean, use $\alpha$ and the normal $(\mu, 1/\sqrt{n})$ distribution of $\bar{X}_n$ to come up with the rejection region.

Consider the case where $\mu_1 > \mu_0$, and fix the likelihood ratio test's level $\alpha$. Then we must have $P_0(\bar{X}_n \text{ is large}) = \alpha$, so the rejection region is

$$
R_L ~=~ \bar{X}_n \geq \mu_0 + \Phi^{-1}(1 - \alpha) \cdot \frac{1}{\sqrt{n}}
$$
where $\Phi$ is the standard normal cdf.

For instance, if $\alpha = 0.05$, then $\Phi^{-1}(0.95) = 1.64$, and the rejection region is $\bar{X_n} > \mu_0 + 1.64 \cdot \displaystyle \frac{1}{\sqrt{n}}$. 

Keep in mind that the critical value $\Phi^{-1}(1 - \alpha) \cdot \frac{1}{\sqrt{n}}$ is not $c$. We never calculated $c$ and nor did we need to. We just had to ensure that the likelihood ratio test has the correct level.

**Method for Carrying Out an LRT:**
- Calculate the likelihood ratio.
- Simplify the expression as much as possible.
- Try to express the condition "likelihood ratio must be small" in terms of only a friendly statistic, not the entire complicated formula for the likelihood ratio.
- Use the null distribution of the friendly statistic to come up with the rejection region of the test, ensuring the desired level. E.g. "The test rejects $H_0$ if the friendly statistic is bigger than 4.1, and $P_0(\text{friendly statistic} > 4.1) = \alpha$."

## Uniformly Most Powerful Tests

As above, let $\mu_1 > \mu_0$. An interesting observation arises:

The value of $\mu_1$ does not affect the rejection region, as long as it's bigger than $\mu_0$.

By the Neyman-Pearson lemma, the likelihood ratio test is the most powerful test of our simple null hypothesis $\mu = \mu_0$ against the simple alternative hypothesis $\mu = \mu_1$ for some fixed $\mu_1 > \mu_0$. 
    
Since the rejection region of our test is the same for all $\mu_1 > \mu_0$, it follows that **the same test is the most powerful for testing $\mu = \mu_0$ vs. $\mu = \mu_1^*$ for any $\mu_1^* > \mu_0$**. In other words, for every fixed value $\mu_1^*$ in the alternative, the same rejection region is the most powerful.

Our test is called *uniformly most powerful* against the composite hypothesis:
- $H_0: \mu = \mu_0$
- $H_1: \mu > \mu_0$

Similarly, for fixed $\mu_1 < \mu_0$, the rejection region of the uniformly most powerful test would be 

$$
\bar{X_n} < \mu_0 - \Phi^{-1}(1-\alpha) \cdot \frac{1}{\sqrt{n}}
$$

Using the same value of $\alpha=0.05$, the null probability of the rejection region is visualized as the highlighted area below.

The rejection region is the same for all $\mu_1 < \mu_0$. The test is therefore *uniformly most powerful* against a composite hypothesis:

- $H_0: \mu = \mu_0$
- $H_1: \mu < \mu_0$

### Concluding Remark
Consider the composite hypothesis $H_1: \mu \neq \mu_0$. The most powerful tests for $\mu < \mu_0$ and $\mu > \mu_0$ have different rejection regions. Without knowing whether $\mu$ is greater than or less than $\mu_0$, we cannot decide which of the two regions to use, so neither of them is uniformly most powerful for $H_1: \mu \neq \mu_0$.

There can be no uniformly most powerful test of $\mu = \mu_0$ versus this alternative $\mu \neq \mu_0$, because if there were such a test, it would beat the tests we just showed were uniformly most powerful against the one-sided alternatives.
