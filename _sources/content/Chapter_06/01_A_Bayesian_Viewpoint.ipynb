# A Bayesian Viewpoint

Let's see what the Benjamini-Hochberg procedure controls. This will motivate us to look at inference through a Bayesian lens as we did at the start of the term.

The FDP $V/R$ is the proportion of false discoveries among all discoveries. It is tempting to say that this ratio estimates "the probability that the null hypothesis is true given that the test made a discovery."

A frequentist would have trouble with the quoted statement because for a frequentist the null hypothesis is either true or not, and there is no randomness associated with its being true or false.

But a Bayesian can talk about the chance that a null hypothesis is true, and any unknown numerical parameter becomes a random variable. This is a rewarding way of looking at inference though it is not without pitfalls. 

Here is a summary of what we have discussed about the Bayesian approach to estimating a numerical parameter.

Start with a prior distribution on the parameter. This distribution represents your uncertainty about the parameter before you have seen any data. After you observe your data, you update your opinion. This updated opinion is represented by the posterior distribution of the parameter given the data.

Sometimes it's useful to have a summary of this updated opinion. One such summary is the *credible interval* you constructed in Week 1. Another is the *highest density interval* you constructed at the same time.

In Data 140 you saw that the mode of the posterior distribution is called the MAP or *maximum a posteriori* estimate of the parameter.

None of these summaries involved the posterior mean. We will end this section by examining how the posterior mean can be used for Bayesian estimation.

## Bayes Estimators
In the frequentist world, let $\theta$ be a fixed numerical parameter that is unknown. Let $\mathbf{X}$ be data whose distribution has $\theta$ as a parameter, and let $\delta(\mathbf{X})$ be a statistic, that is, a numerical function of the data.

As you know from Data 100, a *loss function* $l(\theta, a)$ measures our penalty for using the value $a$ as an estimate of $\theta$. For example, $l(\theta, a) = (a - \theta)^2$ is called *squared error loss*. Another common loss function is the absolute error $l(\theta, a) = \vert a - \theta \vert$, also known as $L^1$ loss. 

The *frequentist risk* in using $\delta(\mathbf{X})$ to estimate $\theta$ is defined by

$$
R(\theta, \delta) ~ = ~ E_\theta l(\theta, \delta(\mathbf{X}))
$$

where $\theta$ is fixed and the expectation is taken over the distribution of $\mathbf{X}$. The frequentist imagines all possible samples including the one at hand, and averages the loss over all of them.

In Data 100 you saw an empirical version of this risk defined as the *average loss*. The estimator $\delta(\mathbf{X})$ was simulated repeatedly, with $Y_i$ denoting the result of the $i$th simulation. The average loss was $\frac{1}{n} \sum_{i=1}^n l(\theta, y_i)$ where $y_i$ is the observed value of $Y_i$.

The risk $R(\theta, \delta)$ is a function of $\theta$ since we averaged over $\mathbf{X}$. In Data 140 you minimized this with respect to $\theta$ to find the best estimate based on squared error loss. In Data 100 you minimized the empirical risk using different loss functions including absolute error.

Set minimization aside and just think of the frequentist risk $R(\theta, \delta)$ as a function of $\theta$. Suppose there is a competing estimator $\gamma(\mathbf{X})$ and you find $R(\theta, \gamma)$. To choose between the two estimators, you would pick the one with the smaller risk.

The choice of estimator is obvious if the risk of one estimator happens to be smaller than the risk of the other for all values of $\theta$. But often it isn't the case that one estimator beats the other for all $\theta$. More typically, one estimator is better over some subset of values of $\theta$, and the other one is better over the complement. 

For the Bayesian, this raises the question of which values they think $\theta$ is likely to have, because they think of $\theta$ as a random variable. For example, if they don't think that $\theta$ is likely to have values in the subset where $\gamma$ beats $\delta$, then they'll just choose $\delta$. 

### Posterior Risk

Now let's work in the Bayesian world where the parameter $\theta$ is a random variable. In Data 140 you became familiar with this approach, especially in the setting of estimating the chance of heads when tossing a [random coin](http://prob140.org/textbook/content/Chapter_21/01_Updating_and_Prediction.html#updating-the-posterior-distribution-of-x-given-s-n). Some notation:

- $f$: the prior density of $\theta$
- $f_{\mathbf{x} \mid \theta}$: the likelihood of the data $\mathbf{x}$ given $\theta$
- $f_{\theta \mid \mathbf{x}}$: the posterior density of $\theta$ given $\mathbf{x}$

As you know, the posterior is proportional to the prior times the likelihood. 

Suppose you have observed the data $\mathbf{x}$. The *posterior risk* of using $\delta(\mathbf{x})$ to estimate $\theta$ is the expected loss under the posterior distribution of $\theta$: 

$$
R_{\rm post}(\theta, \delta(\mathbf{x})) ~ = ~ 
E_{ f_{\theta \mid \mathbf{x}} }l(\theta, \delta(\mathbf{x})) 
$$

Here the data $\mathbf{x}$ have been observed and hence are fixed. The Bayesian now imagines all possible values of $\theta$ and averages over those, using their posterior distribution of $\theta$ given the data.

It's a good idea to compare this definition to the frequentist risk $R(\theta, \delta)$ where $\theta$ is constant and the averaging is over $\mathbf{X}$.

### Bayes Estimator
A *Bayes estimator* or *Bayes action* is the $\delta(\mathbf{x})$ that minimizes the posterior risk.

You know that if you predict a random variable $W$ by the constant $c$, then the minimum absolute error predictor is the median of the distribution of $W$. The least squares preditor is $E(W)$. So if we use squared error loss, then the Bayes esimate is the posterior mean of $\theta$ given $\mathbf{x}$. 

### Beta/Binomial Example
Suppose you are estimating the $p$ of a random coin. Let the prior distribution of this random probability be beta $(r, s)$, and suppose you have observed $S_n$ heads in $n$ tosses of the coin. Then you know that the posterior distribution is beta $(r+S_n, s+(n-S_n))$.

The Bayes estimate of the random probability of heads is the posterior mean. By beta distribution fact, this is

$$
\begin{align*}
\frac{r + S_n}{r + S_n + s+(n-S_n)} ~ &= ~ \frac{r + S_n}{r + s + n} \\
&= ~ \frac{r+s}{r+s+n} \cdot \frac{r}{r+s} ~+~ \frac{n}{r+s+n} \cdot \frac{S_n}{n}
\end{align*}
$$

Notice how the Bayes estimate is a weighted average of the prior mean and the observed proportion of heads. As $n$ gets large, the weight on the prior mean gets smaller and the observed proportion dominates. 

The observed proportion is the MLE of $p$ in the frequentist setting. Therefore the asymptotic distribution of the Bayes estimator is the same as that of of the MLE. Both are asymptotically normal, unbiased, and efficient.

In many Bayesian settings, though not always, the Bayes estimate is a weighted average of the prior mean and the frequentist estimate of the parameter based on the likelihood.
