## Part 1: Efficiency and the Cramér-Rao Bound

We've shown that the MLE is an asymptotically unbiased estimate of the true parameter $\theta$ and that its distribution is approximately $\text{Normal}(\theta, \frac{1}{nI(\theta)})$ as $n$ increases. This is useful because it allows us to make confidence intervals based on the MLE. But could we do better? 

One measure of how good an estimate is relative to another is *efficiency*: 

The *relative efficiency* of unbiased estimators $T_1$ and $T_2$ is defined as $\frac{Var(T_1)}{Var(T_2)}$. $T_1$ is considered a better estimator is this ratio is less than 1.

It's important that we have assumed both estimators to be unbiased. If they had different biases, then we'd have to deal with that before comparing variances.

The **Cramér-Rao theorem** states that for any unbiased estimator $T$ of $\theta$ based on sample size $n$, $Var_{\theta}(T) \geq \frac{1}{nI(\theta)}$.

We say that MLE $\hat{\theta}_n$ is asymptotically efficient because for large $n$ it is almost unbiased and its variance is close to this lower bound. In other words, if we can calculate an MLE, it will be pretty much the best estimate we could get.

### Proof of the Cramér-Rao Bound

Let $\mathbf{X}$ = $X_1, X_2, ... $ be i.i.d random variables with common probability function $f(x \mid \theta)$. Let $T$ be an unbiased estimator of $\theta$.

Define $\displaystyle W(\mathbf{X},\theta) = \sum_{i=1}^{n} s(X_i,\theta)$ where $s$ is the score function defined as in the last seminar by

$$
s(x, \theta) ~ = ~ \frac{d}{d\theta} \log f(x \mid \theta) ~ = ~ \frac{f'(x \mid \theta)}{f(x \mid \theta)} 
$$

From last time we know that $E_\theta(W(\mathbf{X},\theta)) = 0$. We also know that the Fisher Information $I(\theta)$ is defined as $Var(s(X, \theta))$. Since $X_1, X_2, \ldots, X_n$ are i.i.d,

$$
Var_\theta(W(\mathbf{X},\theta)) = nI(θ)
$$

Let $\rho(T, W)$ be the correlation between $T$ and $W(\mathbf{X},\theta)$. So $\rho(T, W) \le 1$. Thus we have

$$
\frac{Cov_\theta(T,W)}{SD_\theta(T)SD_\theta(W)} ≤ 1
$$

which is equivalent to

$$
Var_\theta(T) ≥ \frac{Cov_\theta^2(T,W)}{Var(W)}
$$

by squaring both sides of the second inequality. So

$$
Var_\theta(T) ≥ \frac{Cov_\theta^2(T,W)}{nI(\theta)}
$$

That means the Cramér-Rao theorem would be true if $Cov_\theta(T,W) = 1$. Let's see if we can show this.

Since $T$ is unbiased,

$$
\theta = E_\theta(T) = ∫_\mathbf{x} T(x_1,x_2, ..., x_n) \prod_{i=1}^{n} f(x_i \mid \theta) \; dx_1dx_2...dx_n
$$

Differentiate both sides with respect to $\theta$:

$$
1 = \frac{d }{d\theta} \theta = \frac{d}{d\theta}∫_\mathbf{x} T(x_1,x_2, \ldots, x_n) \prod_{i=1}^{n} f(x_i \mid \theta) \; dx_1dx_2 \ldots dx_n
$$

Under regularity conditions that we assume are met, we can switch the derivative and integal on the right side to get

$$
\begin{align*}
1 &= ∫_\mathbf{x} \frac{d}{d\theta} T(x_1,x_2, \ldots, x_n) \prod_{i=1}^{n} f(x_i \mid \theta) \; dx_1dx_2 \ldots dx_n \\
&= ∫_\mathbf{x} T(x_1,x_2, \ldots, x_n) \frac{d}{d\theta} \prod_{i=1}^{n} f(x_i \mid \theta) \; dx_1dx_2 \ldots dx_n 
\end{align*}
$$

By the product rule for derivatives, $\displaystyle \frac{d}{d\theta} \prod_{i=1}^{n} f(x_i \mid \theta)$ is the sum of $n$ terms, of which term $k$ is equal to

$$
\begin{align*}
f'(x_k \mid \theta) \prod_{i\neq k} f(x_i \mid \theta) 
&= s(x_k, \theta) f(x_k \mid \theta) \prod_{i\neq k} f(x_i \mid \theta) \\
&= s(x_k, \theta) \prod_{i=1}^{n} f(x_i \mid \theta)
\end{align*}
$$

Plug this into our earlier equation to get

$$
1 ~ = ~ ∫_\mathbf{x} T(x_1,x_2, \ldots, x_n)\sum_{i=1}^{n} s(x_i,\theta)\prod_{i=1}^{n} f(x_i \mid \theta) \; dx_1dx_2 \ldots dx_n = E_\theta \left(T\sum_{i=1}^{n} s(X_i,\theta)\right) 
= E_\theta(TW)
$$

Since $E_{\theta}(W) = 0$, we have

$$
1 = E_\theta(TW) = Cov_\theta(T,W)
$$

just as we had hoped. This establishes the Cramér-Rao bound.

## Part 2: Modes of Convergence

Return to the last seminar, which ended with the important result that if $\hat{\theta}_n$ is the MLE of a true numerical parameter $\theta_0$, then under some regularity conditions

$$
\sqrt{n}(\hat{\theta}_n - \theta_0) \rightarrow \text{normal }\left(0, \frac{1}{I(\theta_0)}\right) ~~~~ \text{ as } n \to \infty
$$

We dervied this convergence informally by approximating $\sqrt{n}(\hat{\theta}_n - \theta_0)$ by a ratio: 

$$
\sqrt{n}(\hat{\theta} - \theta_0) \approx \frac{\frac{1}{\sqrt{n}}L'(\theta_0)}{-\frac{1}{n}L''(\theta_0)} ~~~~~~ \text{for large } n
$$
where $L$ is the log likelihood function. 

We showed that the numerator goes to a normal due to the Central Limit Theroem, and the denominator goes to a constant by the Weak Law of Large Numbers. We then concluded that the ratio converges to the normal in the numerator scaled by the constant limit in the denomiator. Now we will make this more precise.

First, we must discuss exactly what we mean by "convergence". As you saw in Data 140, random variables have two kinds of equality: equality as functions and equality in distribution. So also sequences of random variables can converge in various senses, two of which we will examine in this seminar.

### Convergence in Distribution

Consider random variables $X_1, X_2, ...$ not necessarily defined on the same outcome space. Let the cdf of $X_n$ be $F_n$. Let $X$ be another random variable with cdf $F$.


**Definition:** $X_n$ *converges in distribution* to $X$ if $F_n(x) → F(x)$
at all continuity points of the limit cdf $F$.

**Notation:** If $X_n$ converges in distribution to $X$ we will write $X_n \stackrel{d}{\rightarrow} X$.

We will discuss the issue of continuity points shortly. For a start it is fine to focus on the following special case:

**Continuous limit cdf:** If the limit $F$ is continuous, as in the case where the limit distribution is normal, then $X_n \stackrel{d}{\rightarrow} X$ if $F_n(x) → F(x)$ for all $x$. 

Convergence in distribution is a property of the cdf sequence alone. The spaces on which the random variables are defined as functions aren't involved. The next example points out that convergence in distribution is a property of distributions and not of the random variables as functions.

**Example:** Let $Z$ be standard normal. Take $X_n = Z$ for all $n$, and let $X$ be $-Z$. Then for each $n$ we have $F_n = \Phi$, the standard normal cdf. Now the distribution of $X = -Z$ is also standard normal, so $F = \Phi$ as well. Therefore $X_n \stackrel{d}{\rightarrow} X$. That is, the sequence $Z, Z, Z, \dots$ converges in distribution to $-Z$.

#### A Note on Continuity Points

Why must we only look at continuity points of the limit when determining convergence in distribution? Let's take an example to see what the problem is.

Consider the random variable $X_n$ with distribution given by

$$
X_n = 
\begin{cases} 
  \frac{1}{n} & \text{with probability } 1-\frac{1}{n}\\
  1 & \text{with probability } \frac{1}{n}\\
\end{cases}
$$

For large $n$, $X_n$ is close to $0$ with high probability. So it's reasonable to think of the constant $0$ as the limit, or, to be consistent with our notation above, we would like to think of the limit as the random variable $X$ where  $P(X=0) = 1$. 

Now look at the cdf sequence.

$$
F_n(x) = 
\begin{cases}
0 & \text{for } x < \frac{1}{n} \\
1 - \frac{1}{n} & \text{for } \frac{1}{n} \leq x < 1 \\
1 & \text{for } x \geq 1
\end{cases}
$$

$$
F(x) = 
\begin{cases}
0 & \text{for } x < 0 \\
1 & \text{for } x \geq 0
\end{cases}
$$

Draw the graphs of $F_n$ and $F$. For $x < 0$, we have $F_n(x) \to 0 = F(x)$, and for $x > 0$, we have $F_n(x) \to 1 = F(x)$. 

The problem is at $x = 0$, which is the point at which the limit cdf $F$ is discontinuous. 
This point is the only exception to the cdfs' convergence. 

But it is still natural to want to say that $X_n$ converges in distribution to $0$. So the definition chooses to ignore the discontinuity point of the limit, as requiring convergence there is asking too much. 

### Convergence in Probablility

Let $X_1, X_2, \ldots$ and $X$ be random variables defined on the same space.

$X_n$ *converges in probability* to $X$, denoted $X_n \xrightarrow{P} X$, if for every $\epsilon > 0$ we have $P(\vert X_n - X \vert > \epsilon) \rightarrow 0$.

This is the way in which the average of an i.i.d. sample converges to the population mean, as we saw in the Weak Law of Large Numbers in Data 140. 

It is a fact (that we won't prove) that convergence in probability implies convergence in distribution. But convergence in distribution does not imply convergence in probability (see the example below). This is a reason why convergence in distribution is called *weak* convergence.

**Example:** As in our earlier example, let $X_n = Z$ where $Z$ is standard normal, and let $X = -Z$. We showed that $X_n \xrightarrow{d} X$. Now $\vert X_n - X \vert = 2\vert Z \vert$, and for $\epsilon = 2$ we have $P(\vert X_n - X \vert > \epsilon) = P(\vert Z \vert > 1) \approx 0.32 \neq 0$.

### Useful Properties of Convergence


**1. The Continuous Mapping Theorem** (We won't prove this one.)

Let $g$ be a continuous function. Then

$$
\begin{align*}
X_n \xrightarrow{d} X &\implies g(X_n) \xrightarrow{d} g(X)\\
X_n \xrightarrow{P} X &\implies g(X_n) \xrightarrow{P} g(X)
\end{align*}
$$

**2. Sums** (You will prove this one in section.)

$$
X_n \xrightarrow{P} X, Y_n \xrightarrow{P} Y \implies X_n + Y_n \xrightarrow{P} X + Y
$$

An example shows that this result doesn't hold if you replace convergence in probability by convergence in distribution. Inspired by our earlier examples, take $X_n = Y_n = Z$ where $Z$ is standard normal. Take $X = Z$ and $Y = -Z$. Then $X_n \xrightarrow{d} X$ and $Y_n \xrightarrow{d} Y$. Since $X_n + Y_n = 2Z$ for every $n$ whereas $X+Y = 0$, we see that $X_n + Y_n$ does not converge in distribution to $X+Y$.

**3. Slutsky's Theorem** (We won't prove this one.)

Let $c$ be a constant. Then

$$
X_n \xrightarrow{d} X, Y_n \xrightarrow{P} c \implies X_nY_n \xrightarrow{d} cX
$$







Using the Continuous Mapping Theorem and Slutsky's Theorem, we can now explain the convergence of the distribution of the MLE. We have 

$$
\sqrt{n}(\hat{\theta} - \theta_0) \approx \frac{\frac{1}{\sqrt{n}}L'(\theta_0)}{-\frac{1}{n}L''(\theta_0)}
$$

Call the numerator $S_n$. As we saw in the previous seminar, the CLT implies that the numerator $S_n$ converges in distribution to a normal $(0, I(\theta_0))$ random variable.

Recall that the denominator converges to the constant $I(\theta_0)$ in probability, by the WLLN. By the continuous mapping theorem, with $g(x) = \frac{1}{x}$, we obtain a new random variable sequence $T_n = g(-\frac{1}{n}L''(\theta_0))$ that converges in probability to $\frac{1}{I(\theta_0)}$.

Now, we can write the ratio $\sqrt{n}(\hat{\theta} - \theta_0)$ as the product $S_nT_n$. We can apply Slutsky's Theorem to show that this product converges in distribution to the  normal $(0, I(\theta_0))$ random variable times $\frac{1}{I(\theta_0)}$, which is normal $(0, \frac{1}{I(\theta_0)})$.

This method of writing a random sequence as a ratio where the numerator converges in distribution and the denominator converges in probability to a constant is a standard way of establishing convergence in distribution and identifying the limit.
