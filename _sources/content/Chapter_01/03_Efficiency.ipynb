We've shown that the MLE is an asymptotically unbiased estimate of the true parameter $\theta$ and that its distribution is approximately $\text{Normal}(\theta, \frac{1}{nI(\theta)})$ as $n$ increases. This is useful because it allows us to make confidence intervals based on the MLE. But could we do better? 

One measure of how good an estimate is relative to another is *efficiency*: 

The *relative efficiency* of unbiased estimators $T_1$ and $T_2$ is defined as $\frac{Var(T_1)}{Var(T_2)}$. $T_1$ is considered a better estimator is this ratio is less than 1.

It's important that we have assumed both estimators to be unbiased. If they had different biases, then we'd have to deal with that before comparing variances.

The **Cramér-Rao theorem** states that for any unbiased estimator $T$ of $\theta$ based on sample size $n$, $Var_{\theta}(T) \geq \frac{1}{nI(\theta)}$.

We say that MLE $\hat{\theta}_n$ is asymptotically efficient because for large $n$ it is almost unbiased and its variance is close to this lower bound. In other words, if we can calculate an MLE, it will be pretty much the best estimate we could get.

### Proof of the Cramér-Rao Bound

Let $\mathbf{X}$ = $X_1, X_2, ... $ be i.i.d random variables with common probability function $f(x \mid \theta)$. Let $T$ be an unbiased estimator of $\theta$.

Define $\displaystyle W(\mathbf{X},\theta) = \sum_{i=1}^{n} s(X_i,\theta)$ where $s$ is the score function defined as in the last seminar by

$$
s(x, \theta) ~ = ~ \frac{d}{d\theta} \log f(x \mid \theta) ~ = ~ \frac{f'(x \mid \theta)}{f(x \mid \theta)} 
$$

From last time we know that $E_\theta(W(\mathbf{X},\theta)) = 0$. We also know that the Fisher Information $I(\theta)$ is defined as $Var(s(X, \theta))$. Since $X_1, X_2, \ldots, X_n$ are i.i.d,

$$
Var_\theta(W(\mathbf{X},\theta)) = nI(θ)
$$

Let $\rho(T, W)$ be the correlation between $T$ and $W(\mathbf{X},\theta)$. So $\rho(T, W) \le 1$. Thus we have

$$
\frac{Cov_\theta(T,W)}{SD_\theta(T)SD_\theta(W)} ≤ 1
$$

which is equivalent to

$$
Var_\theta(T) ≥ \frac{Cov_\theta^2(T,W)}{Var(W)}
$$

by squaring both sides of the second inequality. So

$$
Var_\theta(T) ≥ \frac{Cov_\theta^2(T,W)}{nI(\theta)}
$$

That means the Cramér-Rao theorem would be true if $Cov_\theta(T,W) = 1$. Let's see if we can show this.

Since $T$ is unbiased,

$$
\theta = E_\theta(T) = ∫_\mathbf{x} T(x_1,x_2, ..., x_n) \prod_{i=1}^{n} f(x_i \mid \theta) \; dx_1dx_2...dx_n
$$

Differentiate both sides with respect to $\theta$:

$$
1 = \frac{d }{d\theta} \theta = \frac{d}{d\theta}∫_\mathbf{x} T(x_1,x_2, \ldots, x_n) \prod_{i=1}^{n} f(x_i \mid \theta) \; dx_1dx_2 \ldots dx_n
$$

Under regularity conditions that we assume are met, we can switch the derivative and integal on the right side to get

$$
\begin{align*}
1 &= ∫_\mathbf{x} \frac{d}{d\theta} T(x_1,x_2, \ldots, x_n) \prod_{i=1}^{n} f(x_i \mid \theta) \; dx_1dx_2 \ldots dx_n \\
&= ∫_\mathbf{x} T(x_1,x_2, \ldots, x_n) \frac{d}{d\theta} \prod_{i=1}^{n} f(x_i \mid \theta) \; dx_1dx_2 \ldots dx_n 
\end{align*}
$$

By the product rule for derivatives, $\displaystyle \frac{d}{d\theta} \prod_{i=1}^{n} f(x_i \mid \theta)$ is the sum of $n$ terms, of which term $k$ is equal to

$$
\begin{align*}
f'(x_k \mid \theta) \prod_{i\neq k} f(x_i \mid \theta) 
&= s(x_k, \theta) f(x_k \mid \theta) \prod_{i\neq k} f(x_i \mid \theta) \\
&= s(x_k, \theta) \prod_{i=1}^{n} f(x_i \mid \theta)
\end{align*}
$$

Plug this into our earlier equation to get

$$
1 ~ = ~ ∫_\mathbf{x} T(x_1,x_2, \ldots, x_n)\sum_{i=1}^{n} s(x_i,\theta)\prod_{i=1}^{n} f(x_i \mid \theta) \; dx_1dx_2 \ldots dx_n = E_\theta \left(T\sum_{i=1}^{n} s(X_i,\theta)\right) 
= E_\theta(TW)
$$

Since $E_{\theta}(W) = 0$, we have

$$
1 = E_\theta(TW) = Cov_\theta(T,W)
$$

just as we had hoped. This establishes the Cramér-Rao bound.
