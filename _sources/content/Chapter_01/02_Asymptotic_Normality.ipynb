## Motivation for the MLE
In Data 140, we introduced the concept of maximum likelihood estimation: the process of producing an estimate for some unknown parameter $\theta$, such that the likelihood of generating the observed data is maximized. Our work was largely motivated by intuition – MLEs were generated for familiar distributions, and gave results that were already known to be good estimates of the unknown parameter. 

Now, we aim to understand why the MLE produces the _best_ estimate of the unknown parameter under fairly general conditions. Further, we seek to establish that if the sample size is large, then the MLE is:
* Consistent, meaning that it is likely to be close to the true parameter
* Approximately normal, with computable mean and variance, which can be used to calculate confidence intervals

## Notation and Assumptions ##
Let $f(x \mid \theta)$ describe the distribution of the random variable $X$ under the fixed, unknown numerical parameter $\theta$. When $X$ is discrete, $f$ represents a probability mass function. When $X$ is continuous, $f$ is a density function. 

The _support_ of $f$ is the set of values over which $f$ is strictly positive.

In our analysis of the MLE, we assume:

- The underlying distribution $f$ has $\theta$ as the only unknown parameter. 
- The support of $f$ is the same regardless of the value of $\theta$. In other words, the possible values of $X$ do not depend on $\theta$. This rules out uniform $[0, \theta]$, for example.
- $f$ is $n$-times differentiable for some arbitrary $n$, though $n=3$ is usually more than we need. That is, we assume that all necessary derivatives of $f$ exist.

**Technical disclaimer:** In some of what follows, we will interchange integrals and derivatives, apply Taylor's theorem to get approximations, and derive limit distributions. But we will not provide the mathematical justifications for all these moves. For that level of detail, take a graduate class.

## The MLE and the Score Function
Consider the i.i.d. sample $X_1, X_2, ..., X_n$ generated under the distribution $f(x \mid \theta)$. 

The likelihood of the data, if $\theta$ is the value of the unknown parameter, is given by
$${\rm Lik}(\theta) = \prod_{i=1}^{n} f(X_i \mid \theta)$$
The log likelihood is
$$L(\theta) = \sum_{i=1}^{n} \log{f(X_i \mid \theta)}$$

The MLE of $\theta$ is the value $\hat{\theta}_n$ that maximizes the log likelihood. **We assume that such a value exists and is unique.**

To determine the MLE of $\theta$, we differentiate the log likelihood with respect to $\theta$, set the derivative equal to 0, and solve for $\theta$. This involves 
$$\frac{d}{d\theta} \sum_{i=1}^{n} \log{f(X_i \mid \theta)}
= \sum_{i=1}^{n} \frac{d}{d\theta}  \log{f(X_i \mid \theta)}$$

Let $X$ be a generic random variable with distribution $f$.
The above equality shows that regardless of the distribution of $X$, our calculation for the MLE of the unknown parameter $\theta$ relies on a sum of $n$ i.i.d. terms: $\frac{d}{d \theta}  \log{f(X_i \mid \theta)}$. This is an encouraging sign because we know a lot about sums of i.i.d. random variables.

This motivates understanding the following real (non-random) function.

Define the **score function** $s(x, \theta)$, as
$$s(x, \theta) = \frac{d}{d\theta} \log{f(x \mid \theta)}$$


### Examples
To familiarize ourselves with the score function, it's helpful to work through two examples, one continuous and one discrete.

#### Normal Mean with Unknown Variance
Consider the random variable $X$ distributed normally with unknown mean $\mu$ and known variance $\sigma^{2}$. 

The score function $s(x, \mu)$ is:
$$s(x, \mu) = \frac{d}{d\mu} \left( \log{\frac{1}{\sqrt{2\pi}\sigma} e^{\frac{-1}{2} (\frac{x-\mu}{\sigma})^{2}}} \right) = \frac{d}{d \mu} \left( \log{\frac{1}{\sqrt{2\pi}\sigma}} - \frac{1}{2} (\frac{x-\mu}{\sigma})^{2} \right) = \frac{x-\mu}{\sigma^{2}}$$

#### Bernoulli Mean
For an example in the discrete case, consider the Bernoulli $(p)$ variable $Y$. The distribution of $Y$ can be written as $p^y(1-p)^{1-y}$ for $y = 0, 1$.

Now, the score function $s(y, p)$ is:
$$s(y, p) = \frac{d}{dp} \left(\log{p^{y}(1-p)^{1-y}}\right) = \frac{d}{dp} \left(y\log{p} + (1-y)\log{(1-p)}\right) = \frac{y}{p} - \frac{1-y}{1-p}$$

Because $y$ can take on only binary values, we can summarize the score function as:
$$s(y, p) = \begin{cases}
  \frac{1}{p-1}  & y = 0 \\
  \frac{1}{p} & y = 1
\end{cases}$$ 

## Mean and Variance of the Score
Now consider the score function applied to a random variable $X$ drawn from $f$. That is, consider the random score $s(X, \theta)$.

### Expectation
Define $E_{\theta}(X)$ as the expectation of $X$ assuming the unknown parameter of the distribution has value $\theta$. This is an important clarification for the work ahead of us: our calculation of expectation _only_ applies for the case where the unknown is equal to $\theta$. 

Let's return to our two examples from above. 

For $X$ with the normal $(\mu, \sigma^2)$ distribution we have
$s(x, \mu) = \frac{x-\mu}{\sigma^{2}}$ Hence $$E_\mu[s(X, \mu)] = \frac{E_\mu[X-\mu]}{\sigma^{2}} = 0$$

When $Y$ has the Bernoulli $(p)$ distribution we have
$$s(y, p) = \begin{cases}
  \frac{1}{p-1}  & y = 0 \\
  \frac{1}{p} & y = 1
\end{cases}$$
Hence $$E_p[s(Y, p)] = (1-p)\cdot\frac{1}{p-1}+p\cdot\frac{1}{p} = 0$$

In both cases, we the expectation is $0$. In fact, this answer generalizes to all distributions.

Note that 
$$s(x, \theta) = \frac{d}{d\theta} \log{f(x \mid \theta)}
= \frac{\frac{d}{d\theta} f(x \mid \theta)}{f(x \mid \theta)} = \frac{f'(x \mid \theta)}{f(x \mid \theta)}$$
where we have expressed $\frac{d}{d\theta} f(x \mid \theta)$ as $f'(x \mid \theta)$ for convenience. Throughout these notes, the $'$ notation will refer to derivatives with respect to the parameter.

Now
$$ E_{\theta}[s(X, \theta)] = \int_{x} \frac{f'(x \mid \theta)}{f(x \mid \theta)}f(x \mid \theta) dx = \int_{x} f'(x \mid \theta)dx = \frac{d}{d\theta} \int_{x} f(x \mid \theta) dx = \frac{d}{d \theta} (1) = 0$$

We have shown that for all $f$, the random score $s(X, \theta)$ has **expectation 0**.

### Variance
Define $Var_{\theta}(X)$ as the variance of a random variable $X$ that has distribution $f(x \mid \theta)$.

We call the variance of the score function the **Fisher information** in a single observation. Formally:

$$I(\theta) := Var_\theta[s(X, \theta)]$$

Since $E_\theta[s(X, \theta)] = 0$ we have

$$I(\theta) = Var_\theta[s(X, \theta)] = E_\theta[s^2(X, \theta)]$$

Let us find the Fisher information, or the variance of the score function, in our two example distributions.

When $X$ has the normal $(\mu, \sigma^2)$ distribution we have
$$Var_\mu[s(X, \mu)] = Var_\mu\big{(} \frac{X-\mu}{\sigma^2} \big{)} = \frac{1}{\sigma^4}Var_\mu(X-\mu) = \frac{1}{\sigma^4}Var(X) = \frac{1}{\sigma^2} = \frac{1}{Var_\mu(X)}$$

When $Y$ has the Bernoulli $(p)$ distribution we have
$$Var_p[s(Y, p)] = E_p[s(Y, p)^{2}] = p\left(\frac{1}{p}\right)^{2} + (1-p)\left(\frac{1}{p-1}\right)^{2} = \frac{1}{p} - \frac{1}{p-1} = \frac{1}{p(1-p)} = \frac{1}{Var(Y)}$$

Once again we notice a pattern in our examples: they both come out to 1 over the variance of the underlying distribution. 

To see if there is a way to generalize this, we will calculate $Var_\theta[s(X, \theta)]$ for $X$ with distribution $f(x \mid \theta)$. 

Fisher's idea was to start with the calculation of $E_{\theta}[s(X, \theta)]$. We know that
$$0 = \int_{x} s(x, \theta) f(x \mid \theta) dx ~~~~ \text{for all } \theta$$
Take the derivative on both sides to see that
$$0 = \frac{d}{d\theta} \int_{x} s(x, \theta) f(x \mid \theta) dx
= \int_{x} \frac{\partial}{\partial \theta} s(x, \theta), f(x \mid \theta) dx$$
Use the product rule of derivatives on the right hand side:
$$
\begin{align*}
0 &= \int_x s'(x, \theta) f(x \mid \theta) dx + \int_x s(x, \theta) f'(x \mid \theta) dx \\
&= E_\theta[s'(X, \theta)] + \int_x \frac{f'(x \mid \theta)}{f(x \mid \theta)} f'(x \mid \theta) dx\\
&= E_\theta[s'(X, \theta)] + \int_{x} \left(\frac{f'(x \mid \theta)}{f(x \mid \theta)}\right)^{2} f(x \mid \theta) dx\\
&= E_{\theta}[s'(X, \theta)] + \int_{x} s^2(x, \theta) f(x \mid \theta) dx\\
&= E_{\theta}[s'(X, \theta)] + E_{\theta}[s^2(X, \theta)]
\end{align*}
$$

Rearranging terms, we arrive at a new expression for Fisher information:
$$I(\theta) = E_{\theta}[s^2(X, \theta)] = -E_{\theta}[s'(X, \theta)]$$

## Consistency
It is still not clear exactly how Fisher Information is connected to the MLE. So let us return to the MLE. 

The maximum likelihood estimator $\hat{\theta}_n$ maximizes the log likelihood $\sum_{i=1}^{n} \log{f(X_i \mid \theta)}$. Equivalently, the MLE maximizes $\frac{1}{n} \sum_{i=1}^{n} \log{f(X_i \mid \theta)}$. 

Thus the random quantity being maximized is the **mean of a sample of i.i.d. $\log{f(X \mid \theta)}$ terms**. 

Denote the true value of the parameter $\theta$ by $\theta_0$. By the Weak Law of Large Numbers, as $n \to \infty$ the sample mean $\frac{1}{n} \sum_{i=1}^{n} \log{f(X_i \mid  \theta)}$ is close to $E_{\theta_0} [\log{f(X \mid \theta)}]$ with high probability. 

Put these facts together:

- The MLE $\hat{\theta}_n$ maximizes $\frac{1}{n} \sum_{i=1}^n \log{f(X_i \mid  \theta)}$. 
- For large $n$, the function being maximized is close to $E_{\theta_0} [\log{f(X \mid \theta)}]$ with high probability. 
- Therefore, for large $n$, the estimator $\hat{\theta}_n$ should be close (with high probability) to the value of $\theta$ that maximizes $E_{\theta_0} [\log{f(X \mid \theta)}]$.

We claim that this maximizing value is the true value $\theta_0$. If the claim is true, then for large $n$ the MLE will be close to the true value of the parameter with high probability. This is desirable in an estimator.

### Establishing the Claim
We will now prove that $\theta_0$ is the value of $\theta$ that maximizes $E_{\theta_0} [\log{f(X \mid \theta)]}$. 

If $\theta_0$ is the maximizing $\theta$, then $E_{\theta_0} \left[\log{f(X \mid \theta_0)}\right]$ should be greater in value than any other $E_{\theta_{0}} \left[\log{f(X \mid \theta)}\right]$. Therefore the difference
$$E_{\theta_0} [\log{f(X \mid \theta)}] - E_{\theta_0} [\log{f(X \mid \theta_{0})}]$$ should be negative.

Notice that
$$E_{\theta_0} [\log{f(X \mid \theta)}] - E_{\theta_0} [\log{f(X \mid \theta_0)}] = E_{\theta_0}[\log{f(X \mid \theta)} - \log{f(X \mid \theta_{0})}] = E_{\theta_0}\left[\log{\frac{f(X \mid \theta)}{f(X \mid \theta_{0})}}\right]$$

We now appeal to **Jensen's Inequality** (see this week's Worksheet) which describes the behavior of expectations of convex or concave functions:
* If $g$ is a concave function, $E[g(X)] \leq g(E[X])$
* If $g$ is a convex function, $E[g(X)] \geq g(E[X])$

Because $\log$ is concave, Jensen's inequality implies that 
$$E_{\theta_0}\left[\log{\frac{f(X \mid \theta)}{f(X \mid \theta_{0})}}\right] \leq \log\left(E_{\theta_0}\left[\frac{f(X \mid \theta)}{f(X \mid \theta_{0})}\right]\right)$$ 

Thus
$$
\begin{align*}
E_{\theta_0} [\log{f(X \mid \theta)}] - E_{\theta_{0}} [\log{f(X \mid \theta_{0}})] &\leq \log{E_{\theta_0}\left[\frac{f(X \mid \theta)}{f(X \mid \theta_{0}}\right]} \\
&= \log{\int_{x} \frac{f(x \mid \theta)}{f(x \mid \theta_{0})} f(x \mid \theta_{0}) dx} \\
&= \log{\int_{x} f(x \mid \theta) dx} = \log(1) = 0
\end{align*}
$$

Thus $E_{\theta_{0}} [\log{f(X \mid \theta)}] \leq 
E_{\theta_0} [\log{f(X \mid \theta_0})]$ for all $\theta$. Hence $E_{\theta_{0}}[\log{f(x \mid \theta)}]$ is maximized by $\theta_{0}$.

Thus for large $n$ the MLE $\hat{\theta}_n$ is close to the true $\theta_0$ with high probability. Estimators that have this property are called *consistent*.

## Asymptotic Normality

We can now identify the asymptotic distribution of the MLE $\hat{\theta}_n$ and see how Fisher Information is involved.

Recall that the log likelihood is
$L(\theta) = \sum_{i=1}^{n} \log{f(X_i \mid \theta)}$. The MLE $\hat{\theta}_n$ satisfies
$0 = L'(\hat{\theta}_n)$.

By consistency, for large $n$ the estimator $\hat{\theta}_n$ is likely to be close to the true $\theta_0$. By a **Taylor expansion** around the point $\theta_0$ we have the following:

For large $n$, with high probability,
$$0 = L'(\hat{\theta}_n) \approx L'(\theta_0) + (\hat{\theta}_n -\theta_0)L''(\theta_0)$$
Thus
$$\hat{\theta}_n - \theta_0 \approx \frac{L'(\theta_0)}{-L''(\theta_0)}$$

Equivalently,
$$\sqrt{n}(\hat{\theta}_n - \theta_0) \approx \frac{\frac{1}{\sqrt{n}}L'(\theta_0)}{-\frac{1}{n}L''(\theta_0)}$$

We will separately examine the denominator and the numerator of the ratio on the right hand side. Recall that for all $\theta$, $L'(\theta) = \sum_{i=1}^n s(X_i, \theta)$.

The denominator is $-\frac{1}{n} \sum_{i=1}^{n} s'(X_i, \theta_0)$. By the Weak Law of Large Numbers, $-\frac{1}{n} \sum_{i=1}^{n} s'(X_{i}, \theta_0)$ is close to 
$-E_{\theta_0}[s'(X, \theta_0)]$ with high probability. That is the Fisher Information $I(\theta_0)$ as we showed earlier.

The numerator is $\frac{1}{\sqrt{n}} \sum_{i=1}^{n} s(X_i, \theta_0)$. By the Central Limit Theorem, for large $n$ the distribution of the sum is approximately normal $(0, nI(\theta_0))$. So for large $n$, the distribution of $\frac{1}{\sqrt{n}} \sum_{i=1}^{n} s(X_{i}, \theta)$ is approximately normal $(0, I(\theta_0))$.

When $n$ is large, the numerator is approximately normal $(0, I(\theta_0))$ and the denominator is close to the constant $I(\theta_0)$. Thus for large $n$, $\sqrt{n}(\hat{\theta_{n}} - \theta_{0})$ is approximately normal $(0, \frac{1}{I(\theta_{0})})$.

Rearranging terms, we have the **asymptotic distribution of the MLE of $\theta_0$**:

- For large $n$, the MLE $\hat{\theta}_n$ is approximately normal $\left(\theta_{0}, \dfrac{1}{nI(\theta_{0})}\right)$.

Notice that when $n$ is large,
- The mean of $\hat{\theta}_n$ is approximately the true parameter $\theta_0$. Thus the MLE is asymptotically unbiased.
- When $I(\theta_0)$ is large, $Var(\hat{\theta}_n)$ is small and thus our estimator is close to the true $\theta_0$. In other words, saying that the uncertainty of the estimator is small is the same as saying that $I(\theta_0)$ is large. This helps explain the term *information*.

We thus have a result that allows us to use the MLE $\hat{\theta}_n$ to construct confidence intervals for the true $\theta_0$.
